```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("netflix_titles.csv")
df.head()
```

```python
# Dataset Information
df.info()
```

```python
# Dataset Shape
print("Rows and Columns:", df.shape)

# Statistical Summary
df.describe(include="all")
```

```python
# Missing Values
df.isnull().sum()
```

```python
import plotly.express as px

# Movies vs TV Shows
type_count = df["type"].value_counts()

fig = px.pie(
    names=type_count.index,
    values=type_count.values,
    title="Movies vs TV Shows"
)

fig.show()
```

```python
country_data = (
    df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

fig = px.bar(
    x=country_data.index,
    y=country_data.values,
    title="Top 10 Countries on Netflix",
    labels={"x": "Country", "y": "Number of Titles"}
)

fig.show()
```

```python
genre_data = (
    df["listed_in"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

fig = px.bar(
    x=genre_data.index,
    y=genre_data.values,
    title="Top 10 Netflix Genres",
    labels={"x": "Genre", "y": "Number of Titles"}
)

fig.show()
```

```python
year_data = df["release_year"].value_counts().sort_index()

fig = px.line(
    x=year_data.index,
    y=year_data.values,
    title="Netflix Content by Release Year",
    labels={"x": "Year", "y": "Number of Titles"}
)

fig.show()
```

```python
rating_data = df["rating"].value_counts()

fig = px.bar(
    x=rating_data.index,
    y=rating_data.values,
    title="Netflix Content Ratings",
    labels={"x": "Rating", "y": "Number of Titles"}
)

fig.show()
```

```python
movies = df[df["type"] == "Movie"].copy()

movies["duration_num"] = (
    movies["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

fig = px.histogram(
    movies,
    x="duration_num",
    nbins=20,
    title="Netflix Movie Duration Distribution",
    labels={"duration_num": "Duration (Minutes)"}
)

fig.show()
```

```python
# Duplicate Values
print("Duplicate Rows:", df.duplicated().sum())

# Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Clean Data
clean_df = df.drop_duplicates().copy()

clean_df["director"] = clean_df["director"].fillna("Unknown")
clean_df["cast"] = clean_df["cast"].fillna("Unknown")
clean_df["country"] = clean_df["country"].fillna("Unknown")
clean_df["rating"] = clean_df["rating"].fillna("Not Rated")

print("\nCleaned Dataset Shape:", clean_df.shape)
```
